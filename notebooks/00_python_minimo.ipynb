## 3. Arreglos de numpy: la forma de guardar una señal

Una señal discreta es una **secuencia de números**. En Python la guardamos en un **arreglo** (`np.array`).

````python
x = np.array([3.0, 5.0, -1.0, 2.0, 4.0])

print("x =", x)
print("número de muestras:", len(x))
````

### Cuidado: lista de Python contra arreglo de numpy

Python también tiene **listas** `[ ]`, pero NO se comportan como señales. Observa la diferencia:

````python
lista = [1, 2, 3]
arreglo = np.array([1, 2, 3])

print("lista * 2   =", lista * 2)      # repite la lista, no multiplica
print("arreglo * 2 =", arreglo * 2)    # multiplica cada muestra: esto es lo que queremos
````

Con arreglos de numpy, las operaciones se aplican **muestra a muestra**:
si `x` y `y` tienen el mismo tamaño, `x + y` suma `x[0]+y[0]`, `x[1]+y[1]`, ...

````python
x = np.array([1.0, 2.0, 3.0])
y = np.array([10.0, 20.0, 30.0])

print("x + y   =", x + y)
print("x * y   =", x * y)       # producto muestra a muestra (NO es producto punto)
print("x ** 2  =", x ** 2)      # cada muestra al cuadrado
print("cos(x)  =", np.cos(x))   # coseno de cada muestra
````

## 4. Índices de Python: empiezan en **0**

Para leer una muestra usamos corchetes: `x[i]`. El **primer** elemento es `x[0]`.

```
posición en Python:   0     1     2     3     4
x               =  [ 3.0,  5.0, -1.0,  2.0,  4.0 ]
                                              ↑
                                    x[-1] = el ÚLTIMO elemento
```

````python
x = np.array([3.0, 5.0, -1.0, 2.0, 4.0])

print("x[0]  =", x[0])     # primer elemento
print("x[1]  =", x[1])     # segundo elemento
print("x[4]  =", x[4])     # quinto (y último) elemento
print("x[-1] =", x[-1])    # índice negativo: cuenta desde el final
````

### Rebanadas (*slicing*): `x[inicio:fin]`

`x[inicio:fin]` toma desde `inicio` **hasta `fin - 1`**: el valor `fin` no se incluye.

````python
print("x[1:4]  =", x[1:4])    # posiciones 1, 2, 3
print("x[:3]   =", x[:3])     # las primeras 3
print("x[-2:]  =", x[-2:])    # las últimas 2
print("x[::2]  =", x[::2])    # una sí, una no (paso de 2)
print("x[::-1] =", x[::-1])   # al revés
````

## 5. La idea más importante: índice de Python ≠ índice de tiempo `n`

En PDS escribimos señales como $x[n]$ donde $n$ puede ser **negativo**. Por ejemplo:

$$x[n] = n^2, \qquad n = -3, -2, \dots, 3$$

Python no sabe nada de eso: sólo guarda los valores **en orden**, empezando en la posición 0.
Por eso, en este curso **siempre guardamos dos arreglos**:

- `n` → los índices de tiempo
- `x` → los valores

````python
n = np.arange(-3, 4)      # -3, -2, -1, 0, 1, 2, 3   (arange NO incluye el 4)
x = n ** 2

print("n =", n)
print("x =", x)
````

Pregunta: **¿cuánto vale $x[n]$ en $n = 0$?** La respuesta correcta es $0^2 = 0$. Pero:

````python
print("x[0] de Python =", x[0], " <- es el valor en n = -3, no en n = 0")
````

Para encontrar el valor en un índice de tiempo dado, **buscamos la posición** donde `n` vale ese número.
Primero con un ciclo (para entender), luego la forma corta:

````python
n_buscado = 0

# Forma 1: recorrer el arreglo n hasta encontrar el valor buscado
for i in range(len(n)):
    if n[i] == n_buscado:
        posicion = i

print(f"n = {n_buscado} está en la posición {posicion} del arreglo")
print(f"x[n = {n_buscado}] = {x[posicion]}")
````

````python
# Forma 2 (corta): n == 0 crea un arreglo de verdadero/falso; x[...] toma donde es verdadero
print("n == 0      ->", n == 0)
print("x[n == 0]   ->", x[n == 0])
````

> **Regla del curso:** cuando una señal no empieza en $n=0$, **nunca** uses `x[i]` pensando que `i` es el tiempo.
> Usa siempre el arreglo `n` para saber a qué instante corresponde cada valor.

## 6. Crear secuencias de números: `np.arange` y `np.linspace`

| Función | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `np.arange(a, b)` | enteros de `a` hasta `b-1` | `np.arange(0, 5)` | `[0 1 2 3 4]` |
| `np.arange(a, b, paso)` | de `a` a `b` (sin incluir) con un paso | `np.arange(0, 1, 0.25)` | `[0. 0.25 0.5 0.75]` |
| `np.linspace(a, b, N)` | `N` puntos de `a` a `b` (**incluye** `b`) | `np.linspace(0, 1, 5)` | `[0. 0.25 0.5 0.75 1.]` |
| `np.zeros(N)` | `N` ceros | `np.zeros(3)` | `[0. 0. 0.]` |
| `np.ones(N)` | `N` unos | `np.ones(3)` | `[1. 1. 1.]` |

````python
print(np.arange(0, 5))
print(np.arange(0, 1, 0.25))
print(np.linspace(0, 1, 5))
print(np.zeros(3))
````

## 7. Ciclos `for`: recorrer una señal muestra por muestra

Muchas operaciones de PDS son **sumas sobre las muestras** (energía, convolución, DFT...).
Un ciclo `for` hace exactamente eso. `range(len(x))` genera las posiciones `0, 1, ..., len(x)-1`.

**Lo que va dentro del ciclo se escribe con sangría (4 espacios).**

````python
x = np.array([3.0, 5.0, -1.0, 2.0, 4.0])

suma = 0.0                       # Paso 1: empezamos el acumulador en cero
for i in range(len(x)):          # Paso 2: para cada posición i = 0, 1, 2, 3, 4
    suma = suma + x[i]           #         sumamos la muestra al acumulador
    print(f"  i = {i}: x[i] = {x[i]:5.1f}  ->  suma parcial = {suma}")

print("Suma total:", suma)
````

numpy tiene funciones que hacen lo mismo en una línea (y más rápido). **Primero entendemos con el ciclo,
después usamos el atajo** — y siempre comprobamos que den lo mismo:

````python
print("Con el ciclo :", suma)
print("Con np.sum   :", np.sum(x))
````

## 8. Funciones: empaquetar un procedimiento para reutilizarlo

Una función recibe **parámetros**, hace un procedimiento y **regresa** (`return`) un resultado.
El texto entre `'''` al inicio es la **documentación** (docstring): explica qué hace.

````python
def valor_medio(x):
    '''
    Calcula el valor medio (promedio) de una señal:
        x_medio = (1/N) * Σ x[n]
    '''
    N = len(x)                 # Paso 1: número de muestras
    suma = 0.0                 # Paso 2: acumulamos todas las muestras
    for i in range(N):
        suma = suma + x[i]
    media = suma / N           # Paso 3: dividimos entre N
    return media


senal = np.array([2.0, 4.0, 6.0, 8.0])   # (evitamos la ñ y los acentos en los nombres)
print("Valor medio:", valor_medio(senal))
````

## 9. Gráficas: `stem` para señales discretas, `plot` para continuas

Una señal discreta **sólo existe en los enteros** $n$; entre muestras **no hay nada**.
Por eso la dibujamos con `stem` (palitos). Si la dibujamos con `plot`, matplotlib une los puntos con líneas
y parece una señal continua: **es un error conceptual frecuente**.

````python
n = np.arange(-5, 6)
x = 0.8 ** np.abs(n)

figura, ejes = plt.subplots(1, 2, figsize=(11, 3))

ejes[0].stem(n, x)
ejes[0].set_title("Correcto: stem (señal discreta)")
ejes[0].set_xlabel("n")

ejes[1].plot(n, x, "o-")
ejes[1].set_title("Engañoso: plot une los puntos")
ejes[1].set_xlabel("n")

plt.tight_layout()
plt.show()
````
